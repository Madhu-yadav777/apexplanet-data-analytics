# ApexPlanet - Exploratory Data Analysis

## Task 1: Foundational Setup & EDA

This notebook performs a comprehensive exploratory analysis of the 
ApexPlanet retail sales dataset.

## Objective

Understand the structure, quality, and key characteristics of the 
retail sales data: missing values, duplicates, distributions, outliers,
 relationships, and business KPIs.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['figure.dpi'] = 100
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['axes.labelsize'] = 11

print('Libraries imported successfully.')
```

## Load Dataset

Load both the raw and the cleaned datasets.

```python
RAW_PATH = '../data/raw/dataset.csv'
CLEAN_PATH = '../data/processed/cleaned_dataset.csv'

raw = pd.read_csv(RAW_PATH)
clean = pd.read_csv(CLEAN_PATH, parse_dates=['Order_Date'])

print(f'Raw shape:   {raw.shape}')
print(f'Clean shape: {clean.shape}')
print(f'Columns: {list(clean.columns)}')
```

## Data Types

Check the data types of every column.

```python
clean.dtypes
```

## Missing Values

Count missing values per column in the cleaned dataset.

```python
print('Missing values per column:')
print(clean.isnull().sum())
print(f'Total missing: {clean.isnull().sum().sum()}')
```

## Descriptive Statistics

Summary statistics for the numerical columns.

```python
clean.describe().T
```

## Key Findings

Compute the headline KPIs from the cleaned dataset.

```python
total_revenue = clean['Total_Sales'].sum()
total_orders = len(clean)
print(f'Total revenue: ${total_revenue:,.2f}')
print(f'Total orders: {total_orders:,}')
print(f'Average order value: ${clean["Total_Sales"].mean():,.2f}')
print(f'Unique customers: {clean["Customer_ID"].nunique():,}')
print(f'Unique products: {clean["Product"].nunique()}')
print(f'Date range: {clean["Order_Date"].min().date()} to {clean["Order_Date"].max().date()}')
```
